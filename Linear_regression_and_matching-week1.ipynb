In this section, I used the dataset `homework_1.1.csv` to study the relationship between the dependent variable `Y` and the predictors `X1`, `X2`, and `X3`.

The main multiple linear regression model was:

\[
Y = \beta_0 + \beta_1 X1 + \beta_2 X2 + \beta_3 X3 + \epsilon
\]

where:

- `Y` is the outcome variable.
- `X1`, `X2`, and `X3` are predictor variables.
- `β0` is the intercept.
- `β1`, `β2`, and `β3` are the regression coefficients.

### 2. Interpretation of the Coefficients

A coefficient tells us how much `Y` is expected to change when a predictor increases by one unit, while the other predictors are held constant.

For example:

- The coefficient of `X1` is about `1.0071`.
- This means that if `X1` increases by one unit, `Y` is expected to increase by approximately `1.0071` units, while `X2` and `X3` are held constant.

Similarly:

- A one-unit increase in `X2` is associated with an increase of about `1.9646` units in `Y`, holding `X1` and `X3` constant.
- A one-unit increase in `X3` is associated with an increase of about `2.9755` units in `Y`, holding `X1` and `X2` constant.

In multiple linear regression, the coefficient of each predictor can also be thought of as its slope while controlling for the other predictors.

### 3. R-squared

The model had:

\[
R^2 = 0.991
\]

This means that approximately 99.1% of the variation in `Y` is explained by `X1`, `X2`, and `X3` together.

This is a very high R-squared value, which indicates that the model explains the data very well.

### 4. Statistical Significance

The p-values for `X1`, `X2`, and `X3` were all approximately `0.000`.

This means that all three predictors are statistically significant in this regression model.

The t-statistic can also be used to compare the strength of evidence for each coefficient.

The t-statistics were:

- `X1`: 60.984

Because `X3` has the largest absolute t-statistic, it is the most statistically significant predictor among the three.

## Part 2: Matching and Effect Estimation

In this section, I used the dataset `homework_1.2.csv` to practice matching observations and estimating an effect after creating more comparable treatment and control groups.

The dataset contains three variables:

- `X`: group/treatment indicator (`X = 1` or `X = 0`)
- `Y`: outcome variable
- `Z`: variable used to determine similarity between observations

The main goal of matching is to compare observations from the `X = 1` group with similar observations from the `X = 0` group based on their values of `Z`.

Instead of simply comparing all observations with `X = 1` against all observations with `X = 0`, matching attempts to create a more comparable control group.

---

### 1. Separating the Treatment and Control Groups

The dataset was first divided into two groups:

- **Treated group:** observations where `X = 1`
- **Control group:** observations where `X = 0`

The variable `Z` was then used as the matching variable.

Conceptually, for an observation in the treatment group with a particular value of `Z`, we want to find one or more control observations with similar values of `Z`.

A smaller difference in `Z` means that the two observations are considered better matches.

---

## Approach A: One Nearest Neighbor

For Questions 4 and 5, I used `NearestNeighbors` with:

`n_neighbors = 1`

This means that for every observation with `X = 1`, the algorithm searches through the observations with `X = 0` and selects the **single closest control observation based on Z**.

For example, if a treated observation has:

`Z = 0.70`

and possible control observations have:

`Z = 0.69, 0.55, 0.40`

then the observation with `Z = 0.69` would be selected because it has the smallest distance from `0.70`.

The matching distance is:

\[
Distance = |Z_{treated} - Z_{control}|
\]

The `kneighbors()` method returned two important outputs:

- `distances`: the distance between each treated observation and its nearest control
- `indices`: the position of the selected control observation

---

### 2. Farthest Nearest-Neighbor Match

After finding the nearest control for every treated observation, I examined all matching distances.

Even though every treated observation received its closest available control, some nearest matches were closer than others.

The **farthest match** is therefore:

\[
\max(\text{all nearest-neighbor distances})
\]

The maximum distance was approximately:

\[
0.210217
\]

Therefore, the answer to Question 4 was:

**0.210217**

This value represents the largest distance in `Z` between a treated observation and its selected nearest control.

---

### 3. Estimating the Effect Using Nearest-Neighbor Matching

For Question 5, I used the matched controls rather than the entire original control group.

This distinction is important.

The goal of matching is to construct a control sample that is more comparable to the treatment sample based on `Z`. Therefore, after performing the matching, the effect should be calculated using the **matched control observations**.

The mean outcome for the treated group was approximately:

\[
\overline{Y}_{treated} = 1.1256
\]

The mean outcome for the matched control sample was approximately:

\[
\overline{Y}_{matched\ control} = 0.5822
\]

The estimated effect was:

\[
Effect =
\overline{Y}_{treated}
-
\overline{Y}_{matched\ control}
\]

Therefore:

\[
Effect = 1.1256 - 0.5822
\]

\[
Effect \approx 0.54336
\]

Therefore, the answer to Question 5 was:

**0.54336**

This means that the average outcome for the treated observations was approximately `0.543` units higher than the average outcome for their matched controls.

An important lesson here is that using the mean of **all controls** would ignore the matching procedure. Once matching has been performed, the matched control sample should be used for this estimator.

---

## Approach B: Radius Matching

For Questions 6 and 7, the matching strategy changed.

Instead of selecting only the single nearest control observation, I selected **all control observations whose Z values were within a distance of 0.2 from each treated observation**.

The matching condition was:

\[
|Z_{treated} - Z_{control}| \leq 0.2
\]

This was implemented using `NearestNeighbors` with:

`radius = 0.2`

This approach is different from one-nearest-neighbor matching.

With nearest-neighbor matching:

> Each treated observation receives one matched control.

With radius matching:

> Each treated observation can receive multiple matched controls, as long as they are within the specified distance.

Therefore, different treated observations can have different numbers of matched controls.

For example:

- Treated observation A may have 15 controls within the radius.
- Treated observation B may have 5 controls.
- Treated observation C may have only 1 control.
- A treated observation may also have no control within the radius.

---

### 4. Understanding Duplicate Matches

The same control observation can fall within the radius of multiple treated observations.

For example:

- Treated A → Controls 2, 5, 8
- Treated B → Controls 5, 7
- Treated C → Controls 5, 9

Control `5` appears three times.

Its first appearance is not considered a duplicate, while the two additional appearances are duplicates.

To calculate the total number of duplicate matches, I combined all matched control indices and calculated:

\[
Duplicates =
Total\ Matched\ Control\ Appearances
-
Unique\ Controls\ Used
\]

The results were:

- Total matched control appearances = **737**
- Unique control observations used = **52**

Therefore:

\[
Duplicates = 737 - 52
\]

\[
Duplicates = 685
\]

Therefore, the answer to Question 6 was:

**685**

This demonstrates that radius matching allows the same control observation to contribute to multiple treatment-control comparisons.

---

### 5. Treatment Observations With No Match

An important issue appeared when calculating the radius-based effect.

There were:

**2 treated observations**

that had no control observations within a distance of `0.2`.

If we try to calculate the mean outcome of an empty neighbor group, Python returns:

`NaN`

which means **Not a Number**.

Including a `NaN` in the final average causes the final result to also become `NaN`.

Therefore, neighbor groups with no available matches cannot contribute a matched-control group mean and must be handled appropriately when calculating the control-side matched averages.

This was an important practical lesson: matching methods do not necessarily guarantee that every treated observation will have an acceptable match.

---

### 6. Estimating the Effect With Radius Matching

Question 7 required a different averaging procedure from simply putting all matched controls into one large sample.

For each treated observation with available neighbors, I first calculated:

\[
\overline{Y}_{control,i}
=
\text{mean Y of the control neighbors for treated observation } i
\]

This creates one control-group mean for each matched neighbor group.

Then, I averaged these neighbor-group means:

\[
\overline{Y}_{neighbor\ groups}
=
\frac{1}{G}
\sum_{i=1}^{G}
\overline{Y}_{control,i}
\]

where `G` represents the number of treated observations with an available neighbor group.

Finally, this average matched-control outcome was compared with the mean outcome of the treated group:

\[
Effect =
\overline{Y}_{treated}
-
\overline{Y}_{neighbor\ groups}
\]

The resulting estimated effect was approximately:

\[
Effect \approx 0.5844
\]

Therefore, the answer to Question 7 was:

**0.5844**

A key point is that the control observations were first averaged **within each neighbor group** before averaging across groups. This prevents treated observations with many control neighbors from automatically receiving more weight simply because they have more matches.

---

## Nearest-Neighbor Matching vs. Radius Matching

| Feature | Nearest Neighbor | Radius Matching |
|---|---|---|
| Matching variable | Z | Z |
| Control candidates | X = 0 | X = 0 |
| Treated observations | X = 1 | X = 1 |
| Matching rule | Closest Z value | All Z values within 0.2 |
| Controls per treated observation | 1 | Potentially many |
| Can a control be reused? | Yes | Yes |
| Can a treated observation have no match? | Generally no when controls exist | Yes |
| Effect obtained in this exercise | 0.54336 | 0.5844 |

## Key Concepts to Remember

### Matching

Matching attempts to compare treated and control observations that are similar according to observed characteristics.

In this exercise, similarity was defined using `Z`.

### Distance

The distance represents how different two observations are with respect to the matching variable.

Because there was only one matching variable:

\[
Distance = |Z_{treated} - Z_{control}|
\]

Smaller distances indicate better matches.

### Nearest-Neighbor Matching

Nearest-neighbor matching selects the closest available comparison observation.

In this exercise, one control was selected for each treated observation.

### Radius Matching

Radius matching selects every comparison observation that falls within a predefined distance.

Here the radius was:

\[
0.2
\]

Therefore, the number of controls could vary across treated observations.

### Duplicate Matches

A control observation may be a suitable match for multiple treated observations. Repeated uses beyond the first occurrence are counted as duplicates for Question 6.

### Matching and Causal Inference

Matching is useful in causal inference because treated and untreated groups may differ systematically.

A simple comparison such as:

\[
Mean(Y|X=1)-Mean(Y|X=0)
\]

may reflect both the treatment relationship and pre-existing differences between the groups.

Matching attempts to make the groups more comparable based on observed characteristics before comparing their outcomes.

However, matching alone does not automatically prove causality. A causal interpretation still depends on assumptions about the study design and whether important confounding variables have been adequately accounted for.

The main lesson from this section is that **how we construct the comparison group matters**.

Nearest-neighbor matching selects the single closest comparison for each treated observation, while radius matching can use multiple sufficiently similar controls.

Because these methods construct different matched comparison samples and weight observations differently, they can produce different effect estimates.
